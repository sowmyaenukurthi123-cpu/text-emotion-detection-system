```python

import re

def detect_emotion(text):

    emotions = {
        "Happy 😊": ["happy", "joy", "excited", "great", "wonderful", "good", "awesome", "glad"],
        "Sad 😢": ["sad", "cry", "lonely", "unhappy", "depressed", "upset", "miss"],
        "Angry 😠": ["angry", "mad", "hate", "furious", "annoyed", "irritated"],
        "Fear 😨": ["afraid", "scared", "fear", "nervous", "worried", "anxious"],
        "Love ❤️": ["love", "care", "affection", "sweet", "beautiful", "dear"],
        "Neutral 😐": []
    }

    words = re.findall(r'\b\w+\b', text.lower())

    scores = {}

    for emotion, keywords in emotions.items():
        scores[emotion] = sum(words.count(word) for word in keywords)

    best_emotion = max(scores, key=scores.get)

    if scores[best_emotion] == 0:
        return "Neutral 😐"

    return best_emotion


# Main Program
print("================================")
print("   TEXT EMOTION DETECTION SYSTEM")
print("================================")

text = input("Enter your sentence: ")

emotion = detect_emotion(text)

print("\nInput Text:", text)
print("Detected Emotion:", emotion)
print("================================")
```
